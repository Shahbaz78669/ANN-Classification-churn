```python
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
```

```python
model=load_model('model.h5')
```

```python
with open('lable_encoder_gender.pkl','rb') as file:
    lable_encoder_gender=pickle.load(file)
    
```

```python
with open('onehot_encoder_geography.pkl','rb') as file:
    onehot_encoder_geo=pickle.load(file)

```

```python
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
# Sample raw input (must match original column names/order before encoding)
sample_input = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 60000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
sample_input_df=pd.DataFrame([sample_input])
sample_input_df
```

```python
geo_encoded=onehot_encoder_geo.transform([[sample_input['Geography']]])
geo_encoded_df=pd.DataFrame(geo_encoded,columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
sample_input_df=pd.DataFrame([sample_input])
sample_input_df
```

```python
sample_input_df['Gender']=lable_encoder_gender.transform(sample_input_df['Gender'])
sample_input_df
```

```python
sample_input_df=pd.concat([sample_input_df.drop("Geography",axis=1),geo_encoded_df],axis=1)
sample_input_df
```

```python
input_scaled=scaler.transform(sample_input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
```

```python
prediction_prob=prediction[0][0]
```

```python
prediction_prob
```

```python
if prediction_prob>0.5:
    print(f"The customer is likely to churn")
else :
    print(f"The customer is likely not to churn")
```
