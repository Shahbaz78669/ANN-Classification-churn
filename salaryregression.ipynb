```python
import pandas as pd
import numpy as np
import pickle
from sklearn.preprocessing import LabelEncoder,OneHotEncoder 
```

```python
data=pd.read_csv(r'Churn_Modelling (1).csv')
data
```

```python
data=data.drop(['RowNumber','CustomerId','Surname'],axis=1)
data
```

```python
lable_encoder=LabelEncoder()
onehot_encoder=OneHotEncoder()
```

```python
data['Gender']=lable_encoder.fit_transform(data['Gender'])
data['Gender']
```

```python
geo_onehot_encoded=onehot_encoder.fit_transform(data[['Geography']]).toarray()
```

```python
geo_onehot_encoder_df = pd.DataFrame(
    geo_onehot_encoded,
    columns=onehot_encoder.get_feature_names_out(['Geography'])
)
```

```python
df=pd.concat([data.drop('Geography',axis=1),geo_onehot_encoder_df],axis=1)
```

```python
df
```

```python
X=df.drop('EstimatedSalary',axis=1)
y=df['EstimatedSalary']
```

```python
with open('lable_encoder_gender.pkl','wb') as file:
    pickle.dump('lable_encoder_gender',file)
with open('onehot_encoder_geography.pkl','wb') as file:
    pickle.dump('onehot_encoder_geography',file)
with open('scaler.pkl','wb') as file:
    pickle.dump('scaler',file)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
X_train
```

ANN Classification

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
model=Sequential([
    Dense(64,activation='relu',input_shape=(X_train.shape[1],)),
    Dense(32,activation='relu'),
    Dense(1)
])
```

```python
model.compile(
    optimizer='adam',
    loss='mean_squared_error',
    metrics=['mae']
)
```

```python
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping,TensorBoard
import datetime
```

```python
tensorflow_callback = tensorflow.keras.callbacks.TensorBoard(log_dir="logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S"), histogram_freq=1)
```

```python
early_stopping_callback=EarlyStopping(monitor="val_loss",patience=10,restore_best_weights=True)
```

```python
history = model.fit(
    X_train,
    y_train,
    validation_data=(X_test, y_test),
    epochs=25,
    callbacks=[tensorflow_callback, early_stopping_callback]
)
```
